```python
# PANDAS
# Pandas is an open source library, providing high-performance, easy-to-use data structures and data analysis tools for Python. 
# The DataFrame is one of Pandas' most important data structures. And is a way to store tabular data
```

```python
#DataFrame from a dictionary
# Pre-defined lists
names = ['United States', 'Australia', 'Japan', 'India', 'Russia', 'Morocco', 'Egypt']
dr =  [True, False, False, False, True, True, True]
cpc = [809, 731, 588, 18, 200, 70, 45]

# Import pandas as pd
import pandas as pd

#Use the pre-defined lists to create a dictionary called my_dict. There should be three key value pairs:
#key 'country' and value names.
#key 'drives_right' and value dr.
#key 'cars_per_cap' and value cpc.

my_dict = {
    'country':names,
    'drives_right':dr,
    'cars_per_cap':cpc
}

# Build a DataFrame cars from my_dict: cars
cars = pd.DataFrame(my_dict)

#print cars
print(cars)
```

```python
# Adding row labels to the DataFrame created

#First you define the row labels
row_labels = ['US','AUS','JPN','IN','RU','MOR','EG']

#Now specify the row labels to cars using the index
cars.index = row_labels

#print cars again
print(cars)
```

```python
# Indexing and selecting data in a csv
# 1. Using square brackets
cars = pd.read_csv("C:/Users/chux/Documents/Python/cars.csv", index_col = 0)

#Print out country column as Pandas Series
print(cars['country'])

# Print out country column as Pandas DataFrame
print(cars[['country']])

#Print out DataFrame with country and drives_right columns
print(cars[['country','drives_right']])
```

```python
# Using square brackets to select rows
 

#print the first 3 rows/observation
print(cars[:3])
#Note: On rows, there is no double brackets.

# Print out fourth, fifth and sixth observation
print(cars[3:6])
```

```python
# 2. Using loc and iloc
#loc is label-based, which means that you have to specify rows and columns based on their row and column labels. 
#iloc is integer index based, so you have to specify rows and columns by their integer index. 

cars = pd.read_csv("C:/Users/chux/Documents/Python/cars.csv", index_col = 0)

# Print out observation for Japan
print(cars.loc[['JPN']])
print(cars.iloc[[2]])

# Print out observations for Australia and Egypt
print(cars.loc[['AUS','EG']])
print(cars.iloc[[1,4]])
```

```python
#Selecting both rows and columns

cars = pd.read_csv("C:/Users/chux/Documents/Python/cars.csv", index_col = 0)
# Print out drives_right value of Morocco
print(cars.loc[['MOR'],['drives_right']])
print(cars.iloc[[5],[2]])

#Print out a sub-DataFrame, containing the observations for Russia and Morocco and the columns country and drives_right.
print(cars.loc[['RU','MOR'],['country','drives_right']])
print(cars.iloc[[4,5],[1,2]])
```

```python
# Print out drives_right column as Series
print(cars.loc[:,'drives_right'])
print(cars.iloc[:,2])

# Print out drives_right column as DataFrame
print(cars.loc[:,['drives_right']])
print(cars.iloc[:,[2]])

# Print out cars_per_cap and drives_right as DataFrame
print(cars.loc[:,['cars_per_cap','drives_right']])
print(cars.iloc[:,[0,2]])
```
