1. While loop
#The while loop is like a repeated if statement. The code is executed over and over again, as long as the condition is True


```python
x = 1
while x < 4 :
    print(x)
    x = x + 1
```

```python
#Code a while loop that keeps running as long as offset is not equal to 0.
# Initialize offset
offset = 8

# Code the while loop
while offset != 0 :
    print("correcting...")
    offset = offset -1
    print(offset)
```

```python
# Adding conditionals to a while loop

 #Inside the while loop, complete the if-else statement:
   #If offset is greater than zero, you should decrease offset by 1.
     #Else, you should increase offset by 1.

offset = -6
while offset != 0:
    print("correting...")
    if offset > 0 :
        offset = offset - 1
    else:
        offset = offset + 1
    print(offset)
    
```

```python
# Loop over a list
# Write a for loop that iterates over all elements of the areas list and prints out every element separately.

areas = [11.25, 18.0, 20.0, 10.75, 9.50]

#code for loop
for area in areas :
    print(area)
```

```python
# Indexes and values
# If you want to access the index information, so where the list element you're iterating over is located, you can use enumerate().

areas = [11.25, 18.0, 20.0, 10.75, 9.50]

for room, area in enumerate(areas) :
    print("room " + str(room)+ " : " + str(area))
```

```python
# For non-programmer folks, room 0: 11.25 is strange. Wouldn't it be better if the count started at 1?

areas = [11.25, 18.0, 20.0, 10.75, 9.50]

for room, area in enumerate(areas, start = 1) :
    print("room " + str(room)+ " : " + str(area))
```

```python
# Loop over list of lists
# Write a for loop that goes through each sublist of house and prints out the x is y sqm, where x is the name of the room and y is the area of the room.

house = [["hallway", 11.25], 
         ["kitchen", 18.0], 
         ["living room", 20.0], 
         ["bedroom", 10.75], 
         ["bathroom", 9.50]]

for room, area in house :
    print("The " + str(room) + " is " + str(area) + " sqm")
    # Alternatives
    #print(room + " is " + str(area))
    #print(room,area)
    
```

```python
# Loop over a dictionary

# The dictionary
europe = {'spain':'madrid', 'france':'paris', 'germany':'berlin',
          'norway':'oslo', 'italy':'rome', 'poland':'warsaw', 'austria':'vienna' }
          
# Iterate over europe
for country, capital in europe.items() :
    print("The capital of " + str(country) + " is " + str(capital))
```

```python
# 2D array
baseball = [[180, 78.4],
            [215, 102.7],
            [210, 98.5],
            [188, 75.2]]

np_baseball = np.array(baseball)

# For loop over np_baseball
for value in np.nditer(np_baseball) :
    print(value)
```

```python
# Loop over DataFrame
# for loop that iterates over the rows of cars and on each iteration perform two print() calls:
# one to print out the row label and one to print out all of the rows contents.

import pandas as pd
cars = pd.read_csv("C:/Users/chux/Documents/Python/cars.csv", index_col = 0)

for lab,row in cars.iterrows():
    print(lab) #These are the row index names
    print(row) #These are the entire detail of each row
```

```python
# Loop over dataframe 2
# The row data that's generated by iterrows() on every run is a Pandas Series. This format is not very convenient to print out.

cars = pd.read_csv("C:/Users/chux/Documents/Python/cars.csv", index_col = 0)

# Using the iterators lab and row, adapt the code in the for loop such that the first iteration prints out "US: 809"
for lab, row in cars.iterrows():
    print(lab + ": " + str(row['cars_per_cap'])) # Adding the str on row since the column contains integers
    #print(lab + ": " + row['country'])  #For a column that has integers, no need of adding str()
```

```python
# Adding a column to a dataframe
# for loop to add a new column, named COUNTRY, that contains a uppercase version of the country names in the "country" column. 

for lab, row in cars.iterrows():
    cars.loc[lab, "COUNTRY"] = row['country'].upper()
print(cars)
```

```python
# Adding a column without using a for loop
# Using the apply() method is easier

cars = pd.read_csv("C:/Users/chux/Documents/Python/cars.csv", index_col = 0)

cars["COUNTRY"] = cars["country"].apply(str.upper)
print(cars)
```
