# Maps
## Takes 2 argument; the function and the iterrable object

```python
countries_list = ['Kenya','Togo','Zambia','South Africa','Djibouti']

for x in countries_list:
    #print(len(x))
    print(f'{x}: {len(x)}')
```

# Instead of the above, you can use the map function as follows

```python
# You add list infront of the map, so that the items will be displyed in a list form

list(map(len,countries_list))
```

# Using map so that it displays the countries name and length

```python
def return_len(x):
    return f'{x}:{len(x)}'

list(map(return_len,countries_list))
```

# Filter
## Takes a function and that function is applied to each element of the iterable object

```python
countries_list = ['Kenya','Togo','Zambia','South Africa',12]

# Say you want to remove the number 12 from the list, you can use the filter function as follows

def is_str(x):
    return type(x) == str

list(filter(is_str, countries_list))
```

# Filter removes the 12 from the countries list
