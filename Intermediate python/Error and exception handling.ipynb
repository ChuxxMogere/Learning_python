```python
try:
    print(100/0)
except:
    print('Something went wrong')
```

# You can also add the error infront of the except as:

```python
try:
    print(100/0)
except ZeroDivisionError:
    print('You cannot divide by 0')
```

# You can add the else block if everything goes well in the try block

```python
try:
    print(100/3)
except ZeroDivisionError:
    print('Something went wrong')
else:
    print('Everything went well')
```

# You can also add the finally block after the whole block.
# This will print regardless of the exception

```python
try:
    print(100/0)
except ZeroDivisionError:
    print('Something went wrong')
else:
    print('Everything went well')
finally:
    print('This will always print')
```

# You also add the type error, and it is always before the else block

```python
try:
    print(100/'hello')
except ZeroDivisionError:
    print('Something went wrong')
except TypeError:
    print('Incompatible types')
else:
    print('Everything went well')
finally:
    print('This will always print')
```
