# When passing arguments in a function, inside the brackets, you specify the number of arguments it is supposed to take i.e

```python
def sum_func(x,y):
    return x + y

# Now, when you call a number  not eqaul to 2, you'll get an error i.e
print(sum_func(3,4))
print(sum_func(3,4,7))
```

# From the above the second is an error, as the function only takes 2 arguments. However, this can be sorted by...

```python
# If no arguments are passed it, it returns 0 as the default values are 0
# However, still this takes 3 arguments

def sum_func(x=0,y=0,z=0):
    return x+y+z

print(sum_func(2,3))
print(sum_func(2,3,4))
print(sum_func(2,3,4,5))
```

# We can solve the above(when more arguments are passed out using the following):

```python
# The asterisk allows someone to pass multliple arguments into the function
# It is not necessarily to use args you can use anything, but conventionally, args is preferred

def sum_func(*args):
    x=0
    for num in args:
        x+=num  # x= x+num
    return x
sum_func(1,2,3,4)
```

# Keywords function(*kwargs). Something that takes keywords and values

```python
def key_word_func(**kwargs):
    print(type(kwargs))
    return kwargs

key_word_func(name = "Linda", age = 34)
```

```python
def key_word_func(**kwargs):
    if 'food' in kwargs:
        return f"My name is {kwargs['name']} and my favourite food is {kwargs['food']}"
    else:
        return f"My name is {kwargs['name']} and I have no favourite food"

key_word_func(name = 'Ian', food = 'Nyamachoma')
```

```python
# If a he doesn't state the favourite food

key_word_func(name = 'Ann')
```

# You can combine args and kwargs together as follows:

```python
# Also note when passing arguements, non-keyword argument should be added be fore the keyword argument
def args_kwargs(*args,**kwargs):
    print(args)
    print(kwargs)

args_kwargs(1,2,3,"random",name= 'Ian', sport = 'hockey')
```
