# Lamda Expressions
## Are a short way of writng a function but only talke simple single line expressions

```python
# Using lambda to see return only string
countries_list = ['Kenya','Togo',45]

#lambda x : type(x) == str
list(filter(lambda x: type(x) == str,countries_list))
```

```python
# Say you have a list, and want to return the multiplication of 2 of that list you can

list_1 = [2,4,6,8,10]

lambda x : x*2
list(map(lambda x:x*2,list_1))
```

```python
# Lambda function that returns a list greater then 10

list_2 = [99,88,2,3,55,66,7,44,10]

list_filtered = list(filter(lambda x:x>10,list_2))
list_filtered
```
