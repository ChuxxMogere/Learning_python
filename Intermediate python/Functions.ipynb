# 2. User Defined functions.
## syntax:
#      def function_name(arguments):
#         block of code to execute when function is called>

```python
# examples of user defined function

def welcome_message():
    print('Welcome home')
```

```python
#calling the function together with the ()

welcome_message()
```

```python
# adding arguments to the function

def welcome_message(x):
    print(f'Welcome {x}')
```

```python
# when calling the function after adding arguments, you MUST SPECIFY the argument, else, error

welcome_message('John')
```

```python
# adding multiple arguments

def welcome_message(x,y,z):
    print(f' Welcome {x}, {y} and {z}')

welcome_message('John', 'Chux', 'Ian')
```

```python
# default arguments incase no arguments are provided

def welcome_message(x = 'name1', y = 'name2', z = 'name3'):
    print(f'Welcome {x}, {y}, and {z}')

welcome_message()
welcome_message('John')
welcome_message(z = 'John')
```

```python
# print vs return function

def print_message():
    print('Hello, this message has been printed')

def return_message():
    return 'Hello, this message has been returned'
```

```python
# Calling print message

print_message()
```

```python
x = print_message()
```

```python
x   # No output because print does not store the value passed
```

```python
# Calling return message

x = return_message()
```

x  # Output is there, becuase return stores the variable x

# Function examples created to solve problems

```python
# Fn that can take a list, set, tuple as a parameter and does the avg,sum

list1 = [3,10,11,3]
avg = sum(list1)/len(list1)
avg
```

```python
def avg(x):
    return sum(x)/len(x)

avg(list1)
```

```python
# adding an exception if an incorrect type is passed

def avg(x):
    if type(x) in [list,set,tuple]:
        return sum(x)/len(x)
    else:
        print('Please pass in either a list, set or a tuple')

avg('hello')
avg(list1)
```

```python
# function that does addition

def add_num(x,y):
    return x + y

add_num(2,3)
```

# User inputs in functions

```python
def create_range_list():
    x = int(input('Please select a start parameter' ))
    y = int(input('Please select a stop parameter' ))
    z = int(input('Please select a step parameter' ))
    return list(range(x,y,z))

create_range_list()
```

# Loops in functions

```python
# Enter any list, and check if it is greater than 10

def greater_than_10(x):
    for element in x:
        if element > 10:
            return True
        else:
            pass        # We write pass instead of return false bcoz, retrun will exit out of the loop, and return the value as an output 
    return False        # return is indented here so that it runs after the loop

greater_than_10([2,34,3])
#greater_than_10([2,4,7])
```

```python
def name_check(list_of_names):
    for name in list_of_names:
        if len(name)>3:
            return True
        else:
            pass
    return False

name_check(['John','Ian','Sam','Willow','Lily',])
#name_check(['Jon','Ian'])
```
