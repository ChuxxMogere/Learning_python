Perform exploratory data analysis on the netflix_data.csv data to understand more about movies from the 1990s decade.

What was the most frequent movie duration in the 1990s? Save an approximate answer as an integer called duration (use 1990 as the decade's start year).

A movie is considered short if it is less than 90 minutes. Count the number of short action movies released in the 1990s and save this integer as short_movie_count.

```python
import pandas as pd

netflix_df = pd.read_csv('C:/Users/chux/Documents/Python/netflix_data.csv', index_col = 0)

# 1. Filtering for movies released in the 90s
movies_90s = netflix_df[
    (netflix_df['type'] == 'Movie') &
    (netflix_df['release_year'] <= 1990) &
    (netflix_df['release_year'] < 2000)
]

# 2. Finding the most frequent duration in that decade
duration = movies_90s['duration'].value_counts().index[0]
duration = int(duration)

print("The most frequent movie duration is: ", duration)

# 3. Counting short action movies
short_movies_90s = movies_90s[movies_90s['duration'] < 90]
short_action_movies = short_movies_90s[short_movies_90s['genre'] == 'Action']
short_movie_count = len(short_action_movies)

print("The number of short action movies is: ",short_movie_count)
```
