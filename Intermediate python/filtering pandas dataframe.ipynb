```python
#Finding all cars that drives right
import pandas as pd
cars = pd.read_csv("C:/Users/chux/Documents/Python/cars.csv", index_col = 0)

#Extract the drives_right column as a Pandas Series and store it as dr.
dr = cars["drives_right"]

#Use dr, a boolean Series, to subset the cars DataFrame. Store the resulting selection in sel
sel = cars[dr] # Where filtering happens. Checks if the rows in dr have true, if not, they are dropped. Works since the column has a boolean series(True/false)

#Print sel, and assert that drives_right is True for all observations
print(sel)

#One line to achieve the above:
sel = cars[cars['drives_right']]
print(sel)
```

```python
#Select the cars_per_cap column from cars as a Pandas Series and store it as cpc.
cpc = cars['cars_per_cap']

#observations that have a cars_per_cap over 500
many_cars = cpc > 500

#Use many_cars to subset cars, similar to what you did before. Store the result as car_maniac.
car_maniac = cars[many_cars]

# Print car_maniac
print(car_maniac)
```

```python
import numpy as np

# Create medium: observations with cars_per_cap between 100 and 500
cpc = cars['cars_per_cap']
between = np.logical_and(cpc > 100, cpc < 500)
medium = cars[between]

# Print medium
print(medium)
```
