```python
# Dice game
#Roll the dice. Use randint() to create the variable dice.
#Finish the if-elif-else construct by replacing ___:
#If dice is 1 or 2, you go one step down.
#if dice is 3, 4 or 5, you go one step up.
#Else, you throw the dice again. The number on the dice is the number of steps you go up.
#Print out dice and step. Given the value of dice, was step updated correctly?

np.random.seed(123)
dice = np.random.randint(1,7)
step = 50

if dice <=2 :
    step = step -1
elif dice <6 :
    step = step +1
else:
    step = step + np.random.randint(1,7)

print(" You've thrown a " +str(dice) + ", the steps you are to take are " + str(step) + "!")
```

```python
# Randow Walk with a for loop
# This code with show the number of steps you'll take based on the dice you are throwing
# If you roll (1,2); you take a step back
# If you roll (3,4,5), you take a step front
#If you roll a 6, you roll the dice again and move forward by that many steps

import numpy as np
np.random.seed(123)

random_walk = [0]

for x in range(100) :
    step = random_walk[-1]  # step is equal to the last element in step
    dice = np.random.randint(0,7)

    if dice <=2 :
        step = step -1
    elif dice <=5 :
        step = step + 1
    else:
        step = step + np.random.randint(0,7)
    random_walk.append(step)

print(random_walk)
```

```python
# Using max, to ensure that step doesn't go below 0

import numpy as np
np.random.seed(123)

random_walk = [0]

for x in range(100):
    step = random_walk[-1]
    dice = np.random.randint(0,7)

    if dice <= 2 :
        step = max(0, step -1)
    elif dice <= 5 :
        step = step + 1
    else :
        step = step + np.random.randint(0,7)
    random_walk.append(step)
print(random_walk)
        
```

```python
# Visulaizing the random walk

import numpy as np
np.random.seed(123)

random_walk = [0]

for x in range(100):
    step = random_walk[-1]
    dice = np.random.randint(0,7)

    if dice <= 2 :
        step = max(0, step -1)
    elif dice <= 5 :
        step = step + 1
    else :
        step = step + np.random.randint(0,7)
    random_walk.append(step)

import matplotlib.pyplot as plt
plt.plot(random_walk)
plt.show()
```

```python
# Simulating multiple walks

import numpy as np
np.random.seed(123)

# Initialize all_walks
all_walks = []

# Simulate random walk five times
for i in range(5) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)

        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        random_walk.append(step)

    # Append random_walk to all_walks
    all_walks.append(random_walk)

# Print all_walks
print(all_walks)
```

```python
# Visualizing all walks

# ...write the above code as is, then followed by

np_aw = np.array(all_walks) # Convert all walks to a numpy array
plt.plot(np_aw)
plt.show()

plt.clf()  #clear the figure

#Transpot the plot so that every row represents the position after 1 throw for 5 random walks
np_aw_t  = np.transpose(np_aw)
plt.plot(np_aw_t)
plt.show()

```

```python
# Add clumsiness incase you fall
# numpy and matplotlib imported, seed set

plt.clf()
# Simulate random walk 20 times
all_walks = []
for i in range(20) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)

        # Implement clumsiness(falling)
        if np.random.rand() <= 0.005 :
            step = 0

        random_walk.append(step)
    all_walks.append(random_walk)

# Create and plot np_aw_t
np_aw_t = np.transpose(np.array(all_walks))
plt.plot(np_aw_t)
plt.show()
```

```python
# Plotting the distribution
# What are the odds that you'll reach 60 steps high

all_walks = []
for i in range(500) :
    random_walk = [0]
    for x in range(100) :
        step = random_walk[-1]
        dice = np.random.randint(1,7)
        if dice <= 2:
            step = max(0, step - 1)
        elif dice <= 5:
            step = step + 1
        else:
            step = step + np.random.randint(1,7)
        if np.random.rand() <= 0.001 :
            step = 0
        random_walk.append(step)
    all_walks.append(random_walk)

# Create and plot np_aw_t
np_aw_t = np.transpose(np.array(all_walks))

# Select last row from np_aw_t: ends.(select the last row. This contains the endpoint of all 500 random walks you've simulated.)
ends = np_aw_t[-1,:]

# Plot histogram of ends, display plot
plt.hist(ends)
plt.show()
```
