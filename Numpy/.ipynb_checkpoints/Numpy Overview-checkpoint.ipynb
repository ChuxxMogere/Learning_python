```python
import numpy as np
```

## Creating Arrays

```python
# Create a list and convert it to an array
list1 = [1,2,3,4,5]

np.array(list1)
```

```python
# You can also assign a value to an array
list2 = [1,34,2,3,4,5]
x = np.array(list2)

type(x)  #To confirm 
np.shape(x)  # 6 to show that there is 6 elements in each row(0 axis)
```

```python
# You can also create an array, by passing on a list

np.array([23,45,67,89])
```

```python
# You can also use a tuple

np.array((2,4,5,8,9))
```

```python
# Creating 2D arrays

arr2d = np.array([[1,2,3],[2,4,5],[5,6,7]])
arr2d
```

```python
arr2d.shape
```

```python
x = [1,2,3,4,5]
y = [6,7,8,9,10]

arr2d_2 = np.array([x,y])
arr2d_2
```

```python
arr2d_2.shape
```
